```python
#part 2.1

#printing original ehr
print(ehr)

#popping patient emma and storing details
og_emma = ehr.pop(0)
print(og_emma)

#converting tuple into a list
list_emma = list(og_emma)
print(list_emma)

#updating the dosage
updated_dosage = '60mg'
list_emma[-1] = updated_dosage
print(list_emma)

#converting list into a tuple
tuple_emma = tuple(list_emma)
print(tuple_emma)

#appending emma's updated dosage into ehr
ehr.append(tuple_emma)
print(ehr)
```

```python
#part 2.2

print(ehr)
#accessing info from dictionary and assigning them into individual variables
patient_1_ID = ehr[0][0]
patient_1_name = ehr[0][1]
patient_1_age = ehr[0][2]
patient_1_med = ehr[0][3]
patient_1_dose =ehr[0][4]

patient_2_ID = ehr[1][0]
patient_2_name = ehr[1][1]
patient_2_age = ehr[1][2]
patient_2_med = ehr[1][3]
patient_2_dose =ehr[1][4]

patient_3_ID = ehr[2][0]
patient_3_name = ehr[2][1]
patient_3_age = ehr[2][2]
patient_3_med = ehr[2][3]
patient_3_dose =ehr[2][4]
```

```python
#part 2.3

#creating dictionary within a dictionary using values stored in individual variables form above
medical_record = {46 : {'name ': patient_1_name,
                        'age': patient_1_age,
                        'med' : patient_1_med,
                        'dose' : patient_1_dose},
                  31 : {'name' : patient_2_name,
                        'age': patient_2_age,
                        'med': patient_2_med,
                        'dose' : patient_2_dose},
                  72 : {'name': patient_3_name,
                        'age' : patient_3_age,
                        'med': patient_3_med,
                        'dose' : patient_3_dose}}
print(medical_record)
```

```python
#question 3:

#creating age variables and making them integers
age_1 = int(medical_record[46]['age'])
age_2 = int(medical_record[31]['age'])
age_3 = int(medical_record[72]['age'])

#mean = sum of observations / no. of observations
age_mean = (age_1 + age_2 + age_3)/ 3
print(age_mean)

#meadian = middle number of sorted numbers
all_ages = [age_1, age_2, age_3]
all_ages.sort()
print(all_ages)
median = all_ages[1]
print(median)

#using built in function
age_max = max([age_1, age_2, age_3])
print(age_max)

#using built in function
age_min = min([age_1, age_2, age_3])
print(age_min)

#min subtracted from max
age_range = age_max - age_min
print(age_range)
```
